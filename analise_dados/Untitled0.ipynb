```python
disciplinas = ["Matematica","Portugues","Ingles","Estatistica"]
df[disciplinas].describe().transpose()
```

```python
# Filtrar os dados de forma a mostrar apenas os alunos com idade superior a 40 anos
df[df["Idade"] > 40]
df.query("Idade > 40").head(5)
mask = df["Idade"] > 40
df[mask].head(5)
```

```python
# Filtram os alunos que realizaram o exame na Escola A
df.query("`Local do exame` == 'Escola A'").head(10)
mask = df["Local do exame"] == "Escola A"
df[mask].head(10)
```

```python
# Filtrar os alunos que são trabalhadores estudante e tem idade inferior a 20
mask01 = df["Trabalhador estudante"] == "Sim"
mask02 = df["Idade"] < 20
df[mask01 & mask02].head(5)
df.query(("`Trabalhador estudante` == 'Sim'") and ("Idade < 20"))
```

```python
# Filtrar os alunos de Lisboa ou Porto
df[df["Concelho"].isin(["Porto","Lisboa"])]
```

```python
# Criar uma nova coluna com a média de cada aluno
df["Media"] = df[disciplinas].mean(axis=1)
df[["Id aluno", "Media"]]
```

```python
# Aprovado ou não (True/False)
df["Aprovado"] = df["Media"] >=10
df[["Id aluno","Media","Aprovado"]].head(10)
```

```python
# Quantos aprovados eu tenho?
df["Aprovado"].sum()
```

```python
df.to_csv("alunos_atualizados.csv")
```

```python
# Coluna classificar
# baseada na coluna da média vai devolver um determinado texto
def classificacao(media):
  if media > 16: return "Excelente"
  elif media < 10: return "Insuficiente"
  return "Bom"

df["Classificar"] = df["Media"].apply(classificacao)
df
```

```python
# Media por concelho
df.groupby("Concelho")["Media"].mean().round(2).reset_index()
```
